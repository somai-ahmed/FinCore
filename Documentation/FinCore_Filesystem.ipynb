## Organisation générale

| Chemin | Fonction |
|---|---|
| `README.md`, `LICENSE`, `.gitignore`, `CHANGELOG.md`, `CONTRIBUTING.md` | Métadonnées du dépôt et documentation relative à la contribution |
| `docs/` | Documentation de l’architecture, du modèle comptable, du moteur de détection, de l’API et des processus |
| `src/` | Ensemble du code source — séparé en `FinCore.Native` (moteur C) et `FinCore.Desktop` (interface VB.NET) |
| `data/` | Données de démonstration, jeux de données de test et fichiers Excel d’exemple |
| `tests/` | Suites de tests du moteur natif et de l’application Desktop |
| `scripts/` | Scripts de compilation, nettoyage, exécution, test et empaquetage |
| `build/` | Sorties de compilation Debug et Release |
| `dist/` | Fichiers finaux destinés à la distribution |
| `.github/workflows/` | Pipelines CI pour la compilation et les tests |

## `src/FinCore.Native/` — Moteur C

```text
FinCore.Native/
├── include/            # En-têtes publics : API, types, erreurs et détection
├── src/
│   ├── api/            # FinCore_ae.c — implémentation de l’API publique C
│   ├── core/           # Gestion des sessions et de la mémoire
│   ├── accounting/     # Comptes, journal, périodes et validation
│   ├── reports/        # Grand livre, balance, bilan et compte de résultat
│   ├── detection/      # Détection, Benford, doublons, nombres ronds et valeurs aberrantes
│   ├── persistence/    # Sérialisation, format FinCore, JSON et CSV
│   └── utils/          # Dates, chaînes de caractères et fonctions mathématiques
├── tests/              # Tests des sessions, comptes, journaux, rapports,
│                       # Benford, détection et persistance
└── CMakeLists.txt
```

Chaque sous-dossier de `src/` correspond à un **module architectural** : Core, Comptabilité, Rapports, Détection, Persistance et Utilitaires. Les modules utilisent des fichiers `.c` associés à leurs fichiers d’en-tête `.h`.

## `src/FinCore.Desktop/` — Interface graphique VB.NET

```text
FinCore.Desktop/
├── FinCore.Desktop.vbproj
├── App.config
├── Core/
│   ├── PInvoke/        # Communication avec la DLL native
│   ├── Models/         # Modèles métier
│   ├── Services/       # Services applicatifs
│   └── Exceptions/     # Gestion des exceptions
├── UI/
│   ├── Forms/          # Fenêtres principales de l’application
│   ├── Controls/       # Contrôles graphiques personnalisés
│   └── Dialogs/        # Boîtes de dialogue
├── Excel/
│   ├── ClosedXmlExporter.vb
│   ├── ClosedXmlImporter.vb
│   ├── Templates/      # Modèles Excel
│   └── Styles/         # Styles et palette graphique
├── Helpers/            # Fonctions utilitaires
└── Resources/          # Icônes et images
```

Cette couche suit l’architecture :

**Interface utilisateur → Services → Modèles → P/Invoke → `FinCore_ae.dll`**

Le **VB.NET** constitue donc la couche applicative et graphique, tandis que le **moteur C** réalise les traitements comptables et les opérations de détection.

## `data/`, `tests/`, `scripts/`, `build/`, `dist/`, `.github/`

```text
data/
├── demo/     # Données de démonstration
├── test/     # Jeux de données pour les tests de détection
└── samples/  # Fichiers Excel d’exemple

tests/
├── Native/   # Tests du moteur C
└── Desktop/  # Tests de l’application VB.NET

scripts/      # Scripts de compilation, nettoyage, exécution, test et empaquetage

build/        # Versions Debug et Release

dist/         # Fichiers finaux prêts à être distribués

.github/workflows/  # Pipelines CI de compilation et de tests
```

- `data/test/` contient des jeux de données correspondant directement aux algorithmes de détection de fraude : **Benford, doublons, nombres ronds et valeurs aberrantes**.
- `dist/` contient uniquement les éléments nécessaires à la distribution finale : l’application graphique compilée, la DLL native, la dépendance ClosedXML et les données de démonstration.
- La CI (`.github/workflows/`) sépare les processus de **compilation** et de **tests**.
