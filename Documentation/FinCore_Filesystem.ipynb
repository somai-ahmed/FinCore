## Top-Level Layout

| Path | Purpose |
|------|---------|
| `README.md`, `LICENSE`, `.gitignore`, `CHANGELOG.md`, `CONTRIBUTING.md` | Standard repo metadata and contribution docs |
| `docs/` | Architecture, accounting model, detection engine, API reference, and process documentation |
| `src/` | All source code — split into `FinCore.Native` (C engine) and `FinCore.Desktop` (VB.NET GUI) |
| `data/` | Demo data, test fixtures for detection algorithms, and sample Excel files |
| `tests/` | Native and Desktop test suites, organized by layer |
| `scripts/` | Batch scripts for build/clean/run/test/package workflows |
| `build/` | Debug/Release build output |
| `dist/` | Final distributable artifacts (EXE, DLLs, demo data) |
| `.github/workflows/` | CI pipelines for build and tests |

## `src/FinCore.Native/` — C Engine

```text
FinCore.Native/
├── include/            # Public headers (API, types, errors, detection)
├── src/
│   ├── api/            # FinCore_ae.c — the public C API implementation
│   ├── core/            # session.c/.h, memory.c/.h
│   ├── accounting/      # accounts, journal, periods, validation
│   ├── reports/         # grand_livre, balance, bilan, compte_resultat
│   ├── detection/       # detection, benford, duplicates, round_numbers, outliers
│   ├── persistence/     # serializer, fincore_format, json, csv
│   └── utils/           # dates, strings, math_utils
├── tests/               # test_session, test_accounts, test_journal, test_reports,
│                        # test_benford, test_detection, test_persistence
└── CMakeLists.txt
```

Each subfolder under `src/` maps to one architectural module (Core, Accounting, Reports, Detection, Persistence, Utils), each with paired `.c`/`.h` files.

## `src/FinCore.Desktop/` — VB.NET GUI

```text
FinCore.Desktop/
├── FinCore.Desktop.vbproj
├── App.config
├── Core/
│   ├── PInvoke/        # FinCoreNative.vb, Structs.vb, Marshaling.vb, NativeConstants.vb
│   ├── Models/         # Account, JournalEntry, JournalLine, Period, DetectionResult, BalanceLine, BilanLine
│   ├── Services/       # Session/Account/Journal/Period/Report/Detection/PersistenceService
│   └── Exceptions/     # FinCoreException, NativeEngineException
├── UI/
│   ├── Forms/          # Login, Main, AccountChart, JournalEntry, GrandLivre, Balance,
│   │                   # Bilan, CompteResultat, Detection, BenfordChart, Import, PeriodManager
│   ├── Controls/       # AmountTextBox, AccountComboBox, DataGridJournal, ReportViewer, DetectionGrid
│   └── Dialogs/        # ExportDialog, DetectionConfigDialog, AboutDialog
├── Excel/
│   ├── ClosedXmlExporter.vb, ClosedXmlImporter.vb
│   ├── Templates/      # GrandLivre, Balance, Bilan, Detection templates
│   └── Styles/         # CellStyles, ColorPalette
├── Helpers/            # DateHelper, CurrencyHelper, ValidationHelper, FileHelper
└── Resources/          # Icons/, Images/
```

This layer follows: **UI → Services → Models → P/Invoke → `FinCore_ae.dll`**.

## `data/`, `tests/`, `scripts/`, `build/`, `dist/`, `.github/`

```text
data/
├── demo/     # demo_data.FinCore, demo_accounts.json, demo_journal.csv
├── test/     # benford_normal.csv, benford_anomaly.csv, duplicates.csv,
│             # outliers.csv, round_numbers.csv
└── samples/  # Ecritures.xlsx

tests/
├── Native/   # Unit/, Integration/
└── Desktop/  # Services/, PInvoke/, Excel/, EndToEnd/

scripts/      # build.bat, clean.bat, run.bat, test.bat, package.bat

build/        # Debug/, Release/

dist/         # FinCoreAccountingGUI.exe, FinCore_ae.dll, ClosedXML.dll, demo_data.FinCore

.github/workflows/  # build.yml, tests.yml
```

- `data/test/` fixtures map 1:1 to the fraud-detection algorithms (Benford, duplicates, round numbers, outliers).
- `dist/` contains only the final shippable artifacts: the compiled GUI, the native DLL, the ClosedXML dependency, and a demo data file.
- CI (`.github/workflows/`) separates build and test pipelines.
