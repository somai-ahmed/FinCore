## Architecture OverviewFinCore follows a **layered native-engine architecture** in which the core accounting and fraud-detection logic is implemented in C and exposed through a stable C API.The VB.NET desktop application acts as the presentation and application layer, communicating with the native engine through **P/Invoke**.```text                       ┌──────────────────────┐                       │    FinCore Desktop   │                       │      VB.NET GUI      │                       └──────────┬───────────┘                                  │                           P/Invoke / ABI                                  │                                  ▼┌──────────────────────────────────────────────────────────────┐│                    FinCore Native Engine                     ││                         C / DLL                              ││                                                              ││  ┌─────────┐ ┌────────────┐ ┌─────────┐ ┌────────────────┐  ││  │  Core   │ │ Accounting │ │ Reports │ │ Fraud Detection│  ││  └─────────┘ └────────────┘ └─────────┘ └────────────────┘  ││       │            │             │              │            ││       └────────────┴─────────────┴──────────────┘            ││                            │                                 ││                     ┌──────▼──────┐                          ││                     │ Persistence │                          ││                     └──────┬──────┘                          │└────────────────────────────┼─────────────────────────────────┘                             │              ┌──────────────┼──────────────┐              ▼              ▼              ▼           .FinCore         JSON           CSV              │              ▼          Application Data        Excel ←──── VB.NET / ClosedXML ────→ Reports```

## 1. Native Engine`FinCore.Native` is the heart of the system. It contains:- Session management- Memory management- Chart of accounts- Journal management- Accounting validation- Fiscal periods- Financial reports- Fraud detection- Persistence- Utility functions### Core Architectural Rule> **Critical accounting and fraud-detection logic belongs in the C engine, not in VB.NET.**VB.NET consumes the engine through its public API.```textVB.NET   │   │ FinCore API   ▼FinCore_ae.dll   │   ├── Session   ├── Accounts   ├── Journal   ├── Validation   ├── Reports   └── Detection```

## 2. Public C API`FinCore_ae.h` is the main public interface of the native engine.It provides access to:```textFinCore API│├── Session API├── Account API├── Journal API├── Period API├── Report API├── Detection API└── Persistence API```Internal implementation details remain inside the native source tree.This creates a clean separation between the **public ABI** and the internal implementation.

## 3. Accounting EngineThe accounting subsystem is organized around:```textAccounts    ↓Journal Entries    ↓Validation    ↓Fiscal Period    ↓Ledger Calculations    ↓Reports```### Journal Entry ValidationA journal entry must be validated before being accepted. Typical validation includes:```textJournal Entry     │     ├── Debit account exists     ├── Credit account exists     ├── Debit = Credit     ├── Valid amount     ├── Valid date     └── Accounting period is open            │            ▼         Accepted```This ensures that invalid accounting data does not propagate into the reporting system.

## 4. Reporting EngineThe reporting subsystem generates:- **Grand livre**- **Balance**- **Bilan**- **Compte de résultat**The reports consume validated accounting data.```text                    Journal                       │                       ▼               Accounting Data                       │          ┌────────────┼────────────┐          ▼            ▼            ▼     Grand Livre    Balance    Financial Data                                     │                           ┌─────────┴─────────┐                           ▼                   ▼                         Bilan          Compte Résultat```> Reports should not independently duplicate accounting calculations.

## 5. Fraud Detection EngineThe detection subsystem contains independent detection algorithms:```textdetection/│├── detection.c├── benford.c├── duplicates.c├── round_numbers.c└── outliers.c````detection.c` acts as the orchestration layer.```text                 Journal / Transactions                          │                          ▼                 Detection Engine                          │          ┌───────────────┼────────────────┐          ▼               ▼                ▼       Benford        Duplicates       Round Numbers          │               │                │          └───────────────┼────────────────┘                          ▼                       Outliers                          │                          ▼                    Risk Evaluation                          │                          ▼                  Detection Results```Each algorithm remains independently testable.

## 6. PersistenceFinCore supports multiple persistence/interoperability formats:```textApplication     │     ▼ Serializer     │     ├── .FinCore     ├── JSON     └── CSV```### .FinCore FormatThe `.FinCore` format is the application's native project/data format.```text.FinCore│├── Metadata├── Chart of Accounts├── Fiscal Periods├── Journal Entries├── Detection Results└── Configuration```JSON and CSV provide interoperability with external systems and tools.

## 7. VB.NET Desktop ArchitectureThe desktop application follows a layered structure:```text┌──────────────────────────────┐│             UI               ││ Forms / Controls / Dialogs   │└──────────────┬───────────────┘               ▼┌──────────────────────────────┐│          Services            ││ Account / Journal / Reports  │└──────────────┬───────────────┘               ▼┌──────────────────────────────┐│           Models             ││ Account / Entry / Results    │└──────────────┬───────────────┘               ▼┌──────────────────────────────┐│          P/Invoke            ││ NativeConstants / Marshaling │└──────────────┬───────────────┘               ▼        FinCore_ae.dll```### UIResponsible for:- User interaction- Forms- Dashboards- Data grids- Reports visualization- Dialogs### ServicesResponsible for application orchestration:- `AccountService`- `JournalService`- `ReportService`- `DetectionService`- `PersistenceService`- `PeriodService`### ModelsRepresent native/application data in VB.NET.### P/InvokeResponsible for communication between VB.NET and the C DLL.

## 8. Excel IntegrationExcel is deliberately kept outside the native accounting engine.### Architecture```text                     Excel                       │             ┌─────────┴─────────┐             ▼                   ▼       Excel Importer      Excel Exporter             │                   ▲             ▼                   │       VB.NET Models        Report Models             │                   ▲             ▼                   │         FinCore API ─────── Reports```### Import Flow```textEcritures.xlsx      ↓ClosedXmlImporter      ↓VB.NET Models      ↓FinCore API      ↓Native Accounting Engine```### Export Flow```textFinCore Engine      ↓Reports      ↓VB.NET Models      ↓ClosedXmlExporter      ↓Excel```> This keeps the C engine independent from Excel and ClosedXML.

## 9. Testing ArchitectureFinCore uses multiple testing levels.```texttests/│├── Native│   ├── Unit│   └── Integration│└── Desktop    ├── Services    ├── PInvoke    ├── Excel    └── EndToEnd```### Native TestsTest the C engine directly:```texttest_sessiontest_accountstest_journaltest_reportstest_benfordtest_detectiontest_persistence```### Desktop TestsTest:```textVB.NET Service      ↓P/Invoke      ↓C DLL      ↓Result```### Excel WorkflowsExcel workflows can be tested independently:```textExcel  ↓Importer  ↓Models  ↓FinCore```End-to-end tests verify the complete application workflow.

## 10. Recommended Accounting FlowThe complete accounting flow is:```textAccount Creation       ↓Journal Entry       ↓Validation       ↓Posting       ↓Ledger       ↓Balance       ↓Financial Statements```The fraud-detection flow operates alongside the accounting system:```textJournal / Transactions        │        ▼ Detection Engine        │        ├── Benford Analysis        ├── Duplicate Detection        ├── Round Number Detection        └── Outlier Detection                 │                 ▼           Risk Evaluation                 │                 ▼          Detection Results```

## 11. Architectural PrinciplesFinCore follows these principles:| Principle | Description ||-----------|-------------|| **Separation of concerns** | Each component has a clearly defined responsibility. || **Native financial engine** | The critical accounting logic is implemented in C. || **Stable API** | The C API provides a controlled interface between the native engine and external applications. || **UI independence** | The accounting engine does not depend on WinForms. || **Excel independence** | The native engine does not depend on Excel or ClosedXML. || **Testability** | Accounting and fraud algorithms are independently testable. || **Interoperability** | The engine can potentially be consumed by applications other than VB.NET. |### Multi-Client Potential```text                 ┌──────────────┐                 │  FinCore API │                 └──────┬───────┘                        │           ┌────────────┼────────────┐           ▼            ▼            ▼        VB.NET         C#         Python        Desktop      Client       Client           │            │            │           └────────────┼────────────┘                        ▼                 FinCore_ae.dll```

Detection Persistence```### Architectural Summary**FinCore Accounting Engine** is a native financial software architecture built around a **C accounting and fraud-detection engine**, exposed through a
